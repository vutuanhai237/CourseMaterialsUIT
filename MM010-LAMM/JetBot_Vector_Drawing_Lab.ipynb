## 1. Nhập đáp án của nhóm

Ví dụ nhóm vẽ số **2**:

Giảng viên chỉ cần sửa `student_points`.

````python

# ============================================================
# GIẢNG VIÊN CHỈ CẦN SỬA PHẦN NÀY
# ============================================================

digit = "2"
student_points = [
    (1, 6),
    (5, 6),
    (4, 4),
    (1, 0),
    (5, 0)
]

# Điểm bắt đầu thực tế của JetBot
start_position = student_points[0]

# Tỷ lệ mô phỏng:
# 1 đơn vị tọa độ = bao nhiêu cm trên mặt sàn
CM_PER_UNIT = 10

# Tốc độ mô phỏng (không ảnh hưởng đến phép tính)
SIMULATION_SPEED = 0.8
````

## 2. Các phép toán vector cơ bản

````python

import math
import matplotlib.pyplot as plt
from IPython.display import display, clear_output
import time

def vector_subtract(A, B):
    # A - B
    return (A[0] - B[0], A[1] - B[1])

def vector_add(A, B):
    return (A[0] + B[0], A[1] + B[1])

def scalar_multiply(k, A):
    return (k * A[0], k * A[1])

def magnitude(A):
    return math.sqrt(A[0]**2 + A[1]**2)

def almost_equal(A, B, tol=1e-9):
    return abs(A[0]-B[0]) < tol and abs(A[1]-B[1]) < tol
````

## 3. Tính vector dịch chuyển từ đáp án của sinh viên

````python

vectors = []

for i in range(len(student_points) - 1):
    current = student_points[i]
    next_point = student_points[i + 1]

    # Vector đi từ current đến next_point:
    v = vector_subtract(next_point, current)
    vectors.append(v)

print(f"Chữ số: {digit}")
print("-" * 55)

for i, v in enumerate(vectors, start=1):
    A = student_points[i-1]
    B = student_points[i]
    print(f"Bước {i}: {A} → {B}")
    print(f"        Vector = {v}")
    print(f"        Độ dài = {magnitude(v):.2f} đơn vị")
    print()
````


## 4. Kiểm tra Vector Addition

Nếu robot bắt đầu tại điểm đầu tiên, tổng tất cả vector phải đưa robot tới điểm cuối:

\[
v_1 + v_2 + \cdots + v_n = P_{final}-P_{start}
\]

Đây là cơ hội để giảng viên hỏi:

> **"Tại sao tổng các vector lại cho chúng ta biết vị trí cuối cùng?"**


````python

total_vector = (0, 0)

for v in vectors:
    total_vector = vector_add(total_vector, v)

expected_total = vector_subtract(
    student_points[-1],
    student_points[0]
)

print("Tổng các vector:")
print("  ", total_vector)

print("\nVector trực tiếp từ điểm đầu đến điểm cuối:")
print("  ", expected_total)

if almost_equal(total_vector, expected_total):
    print("\n✅ ĐÚNG: Tổng vector khớp với chuyển vị từ điểm đầu đến điểm cuối.")
else:
    print("\n❌ CÓ VẤN ĐỀ: Kiểm tra lại danh sách tọa độ.")
````

## 6. Vẽ đường đi của JetBot

````python

def plot_path(points, title="JetBot Vector Drawing"):
    xs = [p[0] for p in points]
    ys = [p[1] for p in points]

    plt.figure(figsize=(8, 8))
    plt.plot(xs, ys, marker="o", linewidth=3)

    # Đánh số các điểm
    for i, (x, y) in enumerate(points):
        label = chr(65 + i) if i < 26 else str(i)
        plt.annotate(
            f" {label} {points[i]}",
            (x, y),
            fontsize=10
        )

    # Vẽ mũi tên cho từng vector
    for i in range(len(points)-1):
        x1, y1 = points[i]
        x2, y2 = points[i+1]

        plt.arrow(
            x1, y1,
            x2-x1, y2-y1,
            length_includes_head=True,
            head_width=0.12,
            alpha=0.6
        )

    plt.axhline(0, linewidth=1)
    plt.axvline(0, linewidth=1)

    margin = 1
    plt.xlim(min(xs)-margin, max(xs)+margin)
    plt.ylim(min(ys)-margin, max(ys)+margin)

    plt.xticks(range(min(xs)-1, max(xs)+2))
    plt.yticks(range(min(ys)-1, max(ys)+2))
    plt.grid(True, alpha=0.3)

    plt.gca().set_aspect("equal", adjustable="box")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.title(title)
    plt.show()

plot_path(
    student_points,
    title=f"JetBot vẽ chữ số {digit}"
)
````


## 7. Mô phỏng JetBot di chuyển từng bước

Mô phỏng này **không điều khiển JetBot thật**. Nó chỉ cho giảng viên xem robot sẽ đi theo đường nào nếu thực hiện đúng các vector.

Mỗi bước:

\[
P_{new}=P_{old}+v
\]


````python

def simulate_jetbot(points, speed=0.8):
    fig, ax = plt.subplots(figsize=(8, 8))

    xs = [p[0] for p in points]
    ys = [p[1] for p in points]

    ax.set_xlim(min(xs)-1, max(xs)+1)
    ax.set_ylim(min(ys)-1, max(ys)+1)
    ax.set_aspect("equal", adjustable="box")
    ax.set_xticks(range(min(xs)-1, max(xs)+2))
    ax.set_yticks(range(min(ys)-1, max(ys)+2))
    ax.grid(True, alpha=0.3)
    ax.set_xlabel("X")
    ax.set_ylabel("Y")
    ax.set_title(f"JetBot Simulation — Digit {digit}")

    # Vẽ toàn bộ nét chữ mờ phía sau
    ax.plot(xs, ys, linestyle="--", alpha=0.25)

    robot, = ax.plot(
        [points[0][0]],
        [points[0][1]],
        marker="o",
        markersize=14
    )

    trail, = ax.plot(
        [points[0][0]],
        [points[0][1]],
        linewidth=3
    )

    for i in range(len(points)):
        ax.annotate(
            chr(65+i) if i < 26 else str(i),
            points[i],
            fontsize=12
        )

    for i in range(1, len(points)):
        x0, y0 = points[i-1]
        x1, y1 = points[i]

        # Nội suy nhiều bước để tạo animation
        n_steps = max(5, int(magnitude((x1-x0, y1-y0)) * 10))

        for t in range(1, n_steps + 1):
            ratio = t / n_steps
            x = x0 + ratio * (x1-x0)
            y = y0 + ratio * (y1-y0)

            robot.set_data([x], [y])

            trail_x = xs[:i]
            trail_y = ys[:i]

            trail_x = trail_x + [x]
            trail_y = trail_y + [y]

            trail.set_data(trail_x, trail_y)

            clear_output(wait=True)
            display(fig)
            time.sleep(0.03 / speed)

    clear_output(wait=True)
    display(fig)

simulate_jetbot(student_points, speed=SIMULATION_SPEED)
````
